```python
from PIL import Image
import requests
from io import BytesIO
from IPython.display import display  # important dans Jupyter/Colab

# 1. Charger l'image depuis GitHub
url = "https://raw.githubusercontent.com/lbreilla-hue/images_SNT/main/image.jpg"
response = requests.get(url)
img = Image.open(BytesIO(response.content))

# 2. Conversion en niveaux de gris
gray_img = img.convert("L")   # "L" = mode 'luminance' = niveaux de gris

# 3. Affichage
print("Image originale :")
display(img)

print("Image en niveaux de gris :")
display(gray_img)
```
